```python
# Emoji Dictionary
dict={
    "😀": "happy",
    "😂": "laugh",
    "❤️": "love",
    "😍": "love",
    "👍": "good",
    "🙏": "thanks",
    "😢": "sad",
    "😭": "cry",
    "😎": "cool",
    "🔥": "awesome",
    "🎉": "celebration",
    "😡": "angry"
}
text_to_emoji={v:k for k,v in dict.items()}
#1
def emoji_to_text(sentence):
    for emoji in dict:
        sentence=sentence.replace(emoji,dict[emoji])
    return sentence
#2
def text_to_emoji_func(sentence):
    words=sentence.split()
    new_sentence=[]
    for word in words:
        if word in text_to_emoji:
            new_sentence.append(text_to_emoji[word])
        else:
            new_sentence.append(word)
    return " ".join(new_sentence)
#3
def remove_emoji(sentence):
    for emoji in dict:
        senetence=sentence.replace(emoji,"")
    return sentence
#4
def count_emoji(sentence):
    count=0
    for emoji in dict:
        count+=sentence.count(emoji)
    return count
    
text = "I love python ❤️🔥😀"

print("Original:", text)
print("Emoji to text:", emoji_to_text(text))
print("Text to emoji:", text_to_emoji_func("I love python awesome happy"))
print("Remove emoji:", remove_emoji(text))
print("Emoji count:", count_emoji(text)) 
```
